```python
import pandas as pd

df = pd.read_csv('../data/ts_per_category.csv')
df['YearMonth'] = pd.to_datetime(df['YearMonth'])
df = df.sort_values('YearMonth')

print(df.shape)
print(df.head())
print(df['Category'].unique())
```

```python
df_furniture       = df[df['Category'] == 'Furniture'][['YearMonth', 'Total_Sales']].set_index('YearMonth')
df_office_supplies = df[df['Category'] == 'Office Supplies'][['YearMonth', 'Total_Sales']].set_index('YearMonth')
df_technology      = df[df['Category'] == 'Technology'][['YearMonth', 'Total_Sales']].set_index('YearMonth')

print("Furniture:", df_furniture.shape)
print("Office Supplies:", df_office_supplies.shape)
print("Technology:", df_technology.shape)
```

```python
from pycaret.time_series import TSForecastingExperiment

exp_furniture = TSForecastingExperiment()
exp_furniture.setup(
    data       = df_furniture,
    target     = 'Total_Sales',
    fh         = 6,           # forecast 6 bulan ke depan
    fold       = 3,
    session_id = 42,
    seasonal_period = 12
)

best_furniture = exp_furniture.compare_models(n_select=3)
print("\n=== TOP 3 MODEL - FURNITURE ===")
print(best_furniture)
```

```python
exp_office = TSForecastingExperiment()
exp_office.setup(
    data       = df_office_supplies,
    target     = 'Total_Sales',
    fh         = 6,
    fold       = 3,
    session_id = 42,
    seasonal_period = 12
)

best_office = exp_office.compare_models(n_select=3)
print("\n=== TOP 3 MODEL - OFFICE SUPPLIES ===")
print(best_office)
```

```python
exp_technology = TSForecastingExperiment()
exp_technology.setup(
    data       = df_technology,
    target     = 'Total_Sales',
    fh         = 6,
    fold       = 3,
    session_id = 42,
    seasonal_period = 12
)

best_technology = exp_technology.compare_models(n_select=3)
print("\n=== TOP 3 MODEL - TECHNOLOGY ===")
print(best_technology)
```

```python
print("=== LEADERBOARD FURNITURE ===")
print(exp_furniture.pull())

print("\n=== LEADERBOARD OFFICE SUPPLIES ===")
print(exp_office.pull())

print("\n=== LEADERBOARD TECHNOLOGY ===")
print(exp_technology.pull())
```

```python
# Ambil model terbaik (ranking pertama dari compare_models)
best_fur  = best_furniture[0]
best_off  = best_office[0]
best_tech = best_technology[0]

print("Model terbaik Furniture      :", type(best_fur).__name__)
print("Model terbaik Office Supplies:", type(best_off).__name__)
print("Model terbaik Technology     :", type(best_tech).__name__)
```

```python
# --- EVALUASI BASELINE DI HOLD-OUT SET ---
print("Evaluasi Baseline Furniture")
holdout_base_fur = exp_furniture.predict_model(best_fur)

print("\nEvaluasi Baseline Office Supplies")
holdout_base_off = exp_office.predict_model(best_off)

print("\nEvaluasi Baseline Technology")
holdout_base_tech = exp_technology.predict_model(best_tech)
```

```python
# --- HYPERPARAMETER TUNING ---
print("Mulai proses tuning (bisa memakan waktu beberapa menit)...")

tuned_fur  = exp_furniture.tune_model(best_fur, optimize='MASE')
tuned_off  = exp_office.tune_model(best_off, optimize='MASE')
tuned_tech = exp_technology.tune_model(best_tech, optimize='MASE')

print("Tuning selesai!")
```

```python
# --- EVALUASI TUNED MODEL DI HOLD-OUT SET ---
print("Evaluasi Tuned Furniture")
holdout_tuned_fur = exp_furniture.predict_model(tuned_fur)

print("\nEvaluasi Tuned Office Supplies")
holdout_tuned_off = exp_office.predict_model(tuned_off)

print("\nEvaluasi Tuned Technology")
holdout_tuned_tech = exp_technology.predict_model(tuned_tech)
```

```python
# Finalize: retrain menggunakan seluruh data (tanpa fold validation)
# Ini wajib sebelum save agar model siap dipakai untuk prediksi nyata
final_fur  = exp_furniture.finalize_model(best_fur)
final_off  = exp_office.finalize_model(best_off)
final_tech = exp_technology.finalize_model(best_tech)

print("\nSemua model berhasil di-finalize!")
```

```python
# Plot forecast vs aktual untuk masing-masing kategori
print("=== Evaluasi Furniture ===")
exp_furniture.plot_model(final_fur, plot='forecast')

print("=== Evaluasi Office Supplies ===")
exp_office.plot_model(final_off, plot='forecast')

print("=== Evaluasi Technology ===")
exp_technology.plot_model(final_tech, plot='forecast')
```

```python
from sqlalchemy import create_engine
import pandas as pd

engine = create_engine(
    "postgresql+psycopg2://postgres:mabaPENS24@localhost:5432/superstore_db"
)

# Simpan model ke file .pkl
exp_furniture.save_model(final_fur,   'model_furniture')
exp_office.save_model(final_off,      'model_office_supplies')
exp_technology.save_model(final_tech, 'model_technology')
print("Model tersimpan sebagai .pkl")

# Ambil metrik dari leaderboard
lb_fur  = exp_furniture.pull()
lb_off  = exp_office.pull()
lb_tech = exp_technology.pull()

def get_metrics(lb):
    row = lb.iloc[0]
    # Kolom metrik PyCaret time series bisa berbeda, cek dengan lb.columns
    mae  = row.get('MAE',  row.get('mae',  None))
    rmse = row.get('RMSE', row.get('rmse', None))
    return float(mae) if mae is not None else None, \
           float(rmse) if rmse is not None else None

# Insert ke tabel ml_predictions
records = []
for category, exp, model, lb in [
    ('Furniture',       exp_furniture, final_fur,  lb_fur),
    ('Office Supplies', exp_office,    final_off,  lb_off),
    ('Technology',      exp_technology, final_tech, lb_tech),
]:
    mae, rmse = get_metrics(lb)
    # Forecast 6 bulan ke depan untuk dapat predicted_value contoh
    forecast = exp.predict_model(model, fh=6)
    
    for i, (idx, row) in enumerate(forecast.iterrows()):
        records.append({
            'prediction_type': 'time_series',
            'category':        category,
            'sub_category':    None,
            'target_year':     int(str(idx)[:4]),
            'target_month':    int(str(idx)[5:7]),
            'predicted_value': float(row.iloc[0]),
            'actual_value':    None,
            'model_name':      type(model).__name__,
            'mlflow_run_id':   None,
            'mae':             mae,
            'rmse':            rmse,
        })

df_pred = pd.DataFrame(records)
df_pred.to_sql('ml_predictions', engine, if_exists='append', index=False)
print(f"Inserted {len(df_pred)} predictions ke ml_predictions")
print(df_pred)
```
